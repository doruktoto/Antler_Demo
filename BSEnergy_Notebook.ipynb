```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv(
    'input_data/Lir_Labs_6_Monate_KW_20260312_converted_renamed.csv',
    usecols=['Date', 'Time', 'Ammonia_AKBVBW1QNH4H01_ival', 'Nitric_oxide_AKBVBW1QNO_H01_ival']
)

# Parse datetime (format: DD.MM.YYYY HH:MM:SS)
df['datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'], format='%d.%m.%Y %H:%M:%S')

# Create DD-HH label for x-axis
df['DD-HH'] = df['datetime'].dt.strftime('%d-%H')

# Convert sensor columns to numeric
df['Ammonia_AKBVBW1QNH4H01_ival'] = pd.to_numeric(df['Ammonia_AKBVBW1QNH4H01_ival'], errors='coerce')
df['Nitric_oxide_AKBVBW1QNO_H01_ival'] = pd.to_numeric(df['Nitric_oxide_AKBVBW1QNO_H01_ival'], errors='coerce')

df = df.sort_values('datetime').reset_index(drop=True)

# Plot with dual y-axes
fig, ax1 = plt.subplots(figsize=(22, 5))

color_nh4 = '#1f77b4'
color_no  = '#d62728'

ax1.plot(df.index, df['Ammonia_AKBVBW1QNH4H01_ival'], color=color_nh4, label='Ammonia (NH4)', linewidth=0.8)
ax1.set_ylabel('Ammonia NH4 (mg/l)', color=color_nh4)
ax1.tick_params(axis='y', labelcolor=color_nh4)

ax2 = ax1.twinx()
ax2.plot(df.index, df['Nitric_oxide_AKBVBW1QNO_H01_ival'], color=color_no, label='Nitric Oxide (NO)', linewidth=0.8)
ax2.set_ylabel('Nitric Oxide NO (mg/l)', color=color_no)
ax2.tick_params(axis='y', labelcolor=color_no)

# X-axis: one tick per day (every 24 hourly rows)
tick_step = 24
tick_indices = df.index[::tick_step]
tick_labels  = df['DD-HH'].iloc[::tick_step]
ax1.set_xticks(tick_indices)
ax1.set_xticklabels(tick_labels, rotation=45, ha='right', fontsize=7)

ax1.set_xlabel('Day-Hour (DD-HH)')
ax1.set_title('Ammonia & Nitric Oxide over 6 Months (Lir Labs) — by Day-Hour')

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper right')

plt.tight_layout()
plt.show()
```
